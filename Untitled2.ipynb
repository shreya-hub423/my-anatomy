```python
import os
import matplotlib.pyplot as plt
import pandas as pd

# 1. Load the dataset using the path returned by kagglehub
# path variable from your kagglehub output:
csv_file = os.path.join(path, "covid_19_india.csv")
df = pd.read_csv(csv_file)

# 2. Inspect the dataset
print("--- DATASET INFO ---")
df.info()

print("\n--- DATASET DESCRIBE ---")
print(df.describe())

# 3. Clean & Calculate Death Rate and Recovery Rate (%)
# Convert Date column to datetime format
df["Date"] = pd.to_datetime(df["Date"], format="%Y-%m-%d")

# Ensure numeric types for calculation
df["Confirmed"] = pd.to_numeric(df["Confirmed"], errors="coerce")
df["Cured"] = pd.to_numeric(df["Cured"], errors="coerce")
df["Deaths"] = pd.to_numeric(df["Deaths"], errors="coerce")

# Calculate Death Rate (%) and Recovery Rate (%)
df["Death_Rate"] = (df["Deaths"] / df["Confirmed"]) * 100
df["Recovery_Rate"] = (df["Cured"] / df["Confirmed"]) * 100

# Group by Date to get daily national trends
daily_df = (
    df.groupby("Date")[["Confirmed", "Cured", "Deaths"]].sum().reset_index()
)
daily_df["Death_Rate"] = (daily_df["Deaths"] / daily_df["Confirmed"]) * 100
daily_df["Recovery_Rate"] = (daily_df["Cured"] / daily_df["Confirmed"]) * 100

# 4. Plot 1: Line Chart (Death Rate vs Recovery Rate over Time)
plt.figure(figsize=(12, 6))
plt.plot(
    daily_df["Date"],
    daily_df["Death_Rate"],
    label="Death Rate (%)",
    color="red",
    linewidth=2,
)
plt.plot(
    daily_df["Date"],
    daily_df["Recovery_Rate"],
    label="Recovery Rate (%)",
    color="green",
    linewidth=2,
)
plt.title("COVID-19 India: Death Rate vs Recovery Rate Over Time")
plt.xlabel("Date")
plt.ylabel("Percentage (%)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()

# 5. Plot 2: Bar Chart (State-wise Average Death Rate vs Recovery Rate - Top 10 States)
state_df = (
    df.groupby("State/UnionTerritory")[["Confirmed", "Cured", "Deaths"]]
    .max()
    .reset_index()
)
state_df["Death_Rate"] = (state_df["Deaths"] / state_df["Confirmed"]) * 100
state_df["Recovery_Rate"] = (
    state_df["Cured"] / state_df["Confirmed"]
) * 100

# Filter top 10 states by total confirmed cases for a clean visualization
top10_states = state_df.nlargest(10, "Confirmed")

top10_states.plot(
    x="State/UnionTerritory",
    y=["Death_Rate", "Recovery_Rate"],
    kind="bar",
    figsize=(12, 6),
    color=["crimson", "mediumseagreen"],
)
plt.title(
    "Death Rate vs Recovery Rate Across Top 10 Affected States/UTs in India"
)
plt.xlabel("State / Union Territory")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```
